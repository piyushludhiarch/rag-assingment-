# Spare Parts Assistant Agent

Send a question to the assistant and display the selected tool and answer.

Run the cells from top to bottom.

```python
from pathlib import Path
import sys

PROJECT = Path.cwd().resolve()
if PROJECT.name.lower() == 'notebooks':
    PROJECT = PROJECT.parent
sys.path.insert(0, str(PROJECT))
%cd $PROJECT
print('Project folder:', PROJECT)
```

```python
from ipywidgets import Text, Button, VBox, Output
from agent import run_agent

question_box = Text(
    value="The pump is leaking near the seal. Which part do I need?",
    description="Question:",
    placeholder="Type a maintenance question",
    layout={"width": "700px"},
)
ask_button = Button(
    description="Ask assistant",
    button_style="primary",
    icon="search",
)
answer_output = Output()

def ask_assistant(button):
    question = question_box.value.strip()
    answer_output.clear_output(wait=True)
    if not question:
        with answer_output:
            print("Please type a question.")
        return
    with answer_output:
        result = run_agent(question)
        print("Selected tool:", result["tool_used"])
        print(result["answer"])

ask_button.on_click(ask_assistant)
display(VBox([question_box, ask_button, answer_output]))
```
