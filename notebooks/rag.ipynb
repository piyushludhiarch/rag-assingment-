```python
from pathlib import Path
import sys

PROJECT = Path.cwd().resolve()
if PROJECT.name.lower() == 'notebooks':
    PROJECT = PROJECT.parent
sys.path.insert(0, str(PROJECT))
%cd $PROJECT
print('Project folder:', PROJECT)
```

```python
from ipywidgets import Text, Button, VBox, Output
from rag import GuideRAG

guide_rag = GuideRAG()
question_box = Text(
    value="How do I replace a pump seal safely?",
    description="Question:",
    placeholder="Search the maintenance guide",
    layout={"width": "700px"},
)
search_button = Button(
    description="Search guide",
    button_style="primary",
    icon="search",
)
answer_output = Output()

def search_guide(button):
    question = question_box.value.strip()
    answer_output.clear_output(wait=True)
    if not question:
        with answer_output:
            print("Please type a question.")
        return
    with answer_output:
        result = guide_rag.answer(question)
        print("Retrieved chunks:")
        for chunk in result["sources"]:
            print("-", chunk[:200])
        print("\nAnswer:")
        print(result["answer"])

search_button.on_click(search_guide)
display(VBox([question_box, search_button, answer_output]))
```
